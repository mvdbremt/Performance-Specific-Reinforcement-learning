```python
import numpy as np
import time
import matplotlib
import matplotlib.pyplot as plt
import random 
import numpy as np
```

```python
goaldistr=np.random.normal(loc=130,scale=10,size=1000)
```

```python
def KL(a,b):
       
        if len(a)<len(b):
            return 1
        return np.sum(a*np.log(a/b))
```

```python
klvalues=[]
for i in range(60,180):
    klvalues.append(KL(goaldistr,np.random.normal(loc=i,scale=10,size=1000)))
    if klvalues[-1]<=0:
        print(len(klvalues),i)
```

```python
plt.plot(klvalues)
```
