```python
def changeReward(reward,done,num,goal):
    if done:
        if num>goal:
             return reward-0.01*(num-goal)#+10
        return reward-0.01*(goal-num)#-10
    else:
        if num>goal:
             return -(reward+0.01*(num-goal))
        return 1 + 0.01 * num
```

```python
x=[]
xdone=[]
for i in range( 200):
    x.append(changeReward(1,False,i,100))
    xdone.append(changeReward(1,True,i,100))
plt.plot(x,color="r")
plt.plot(xdone)
```
