```python
import matplotlib.pyplot as plt
```

```python
goal = 130
```

```python
%store -r goal1LearnerEp
Goal1episodesResultswithoutOutliers=[]
for i in goal1LearnerEp:
    if i >goal * 0.1 and i < goal*2:
        Goal1episodesResultswithoutOutliers.append(i)

print(len(goal1LearnerEp)-len(Goal1episodesResultswithoutOutliers))
```

```python
%store -r goal1learnerTestEp
%store -r goal2learnerTestEp
%store -r goal3learnerTestEp
%store -r goal4learnerTestEp

plt.plot(goal1learnerTestEp)
plt.plot(goal2learnerTestEp)
plt.plot(goal3learnerTestEp)
plt.plot(goal4learnerTestEp)
plt.axhline(y= goal,color ='g')
plt.legend(["Simple reward","Increased reward","Negative reward","Reduced reward"])
```

```python
previous100AveragePlot=[[],[],[],[]]
for i in range(1,60):
    previous100AveragePlot[0].append(sum(goal1LearnerEp[(i*500)-100:(i*500)])/100)
    previous100AveragePlot[1].append(sum(goal2LearnerEp[(i*500)-100:(i*500)])/100)
    previous100AveragePlot[2].append(sum(goal3LearnerEp[(i*500)-100:(i*500)])/100)
    previous100AveragePlot[3].append(sum(goal4LearnerEp[(i*500)-100:(i*500)])/100)

plt.plot(previous100AveragePlot[0])
plt.plot(previous100AveragePlot[1])
plt.plot(previous100AveragePlot[2])
plt.plot(previous100AveragePlot[3])
plt.legend(["Simple reward","Increased reward","Negative reward","Reduced reward"])
plt.axhline(y= goal,color ='g')
```
