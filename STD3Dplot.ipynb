```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
goal = 130
std_range = 60
i_range = goal * 3

# Initialize arrays
X = []
Y = []
Z = []

# Generate values
for std in range(std_range):
    for i in range(i_range):
        if i < goal - std:
            value = 1
        elif i > goal + std:
            value = -1
        else:
            value = goal - abs(goal - i)

        # Store values for plotting
        X.append(std)
        Y.append(i)
        Z.append(value)

# Convert lists to numpy arrays
X = np.array(X)
Y = np.array(Y)
Z = np.array(Z)

# Reshape for surface plot
X = X.reshape(std_range, i_range)
Y = Y.reshape(std_range, i_range)
Z = Z.reshape(std_range, i_range)

# Plotting the 3D surface
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

# Labels
ax.set_xlabel('s variable')
ax.set_ylabel('Agent s episode result')
ax.set_zlabel('Reward')
ax.set_title('3D Plot of rewards per result depending on given s value')

plt.show()
plt.savefig("SVG/GoalRewardSTD.svg")
```
